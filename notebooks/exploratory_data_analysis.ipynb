# House Price Prediction — Exploratory Data Analysis
This notebook inspects the California Housing dataset and creates basic visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_preprocessing import load_dataset
X, y, feature_names = load_dataset()
df = X.copy()
df['MedHouseVal'] = y
display(df.head())
display(df.describe().T)
```

```python
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
print('\nData types:\n', df.dtypes)
```

```python
sns.histplot(df['MedHouseVal'], kde=True)
plt.title('Distribution of Median House Values')
plt.xlabel('Median house value (units of $100,000)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm', center=0)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```
